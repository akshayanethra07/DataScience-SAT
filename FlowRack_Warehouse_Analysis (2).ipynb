```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

orders = pd.read_csv('/content/orders.csv')
picks = pd.read_csv('/content/picks.csv')
items = pd.read_csv('/content/items.csv')
workers = pd.read_csv('/content/workers.csv')
zones = pd.read_csv('/content/zones.csv')

print("All five datasets loaded successfully!")
```

```python

datasets = {
    "Orders": orders,
    "Picks": picks,
    "Items": items,
    "Workers": workers,
    "Zones": zones
}

for name, df in datasets.items():
    print(f"\n--- {name} ---")
    print("Rows and columns:", df.shape)
    print("Column names:", df.columns.tolist())
    display(df.head())
```

```python

# Clean the picks dataset
picks_clean = picks.copy()

picks_clean = picks_clean.drop_duplicates()

# Standardize zone names
picks_clean['zone'] = (
    picks_clean['zone']
    .astype('string')
    .str.strip()
    .str.upper()
    .str.replace(r'[\s_-]+', '', regex=True)
    .map({
        'ZONEA': 'Zone-A',
        'ZONEB': 'Zone-B',
        'ZONEC': 'Zone-C',
        'ZONED': 'Zone-D',
        'ZONEE': 'Zone-E'
    })
)

# Convert timestamps
picks_clean['start'] = pd.to_datetime(
    picks_clean['start'], errors='coerce'
)
picks_clean['finish'] = pd.to_datetime(
    picks_clean['finish'], errors='coerce'
)

# Calculate duration in minutes
picks_clean['picking_minutes'] = (
    picks_clean['finish'] - picks_clean['start']
).dt.total_seconds() / 60

print("Cleaned picks:", picks_clean.shape)
print("Missing finish times:",
      picks_clean['finish'].isna().sum())
print("Negative durations:",
      (picks_clean['picking_minutes'] < 0).sum())

display(picks_clean.head())
```

```python

workers_clean = workers.drop_duplicates().copy()
zones_clean = zones.drop_duplicates().copy()

# Fill missing worker experience with Unknown
workers_clean['experience_band'] = (
    workers_clean['experience_band'].fillna('Unknown')
)

# Standardize zone labels in the zone lookup table
zones_clean['zone'] = (
    zones_clean['zone']
    .astype(str)
    .str.strip()
    .str.title()
)

analysis = picks_clean.merge(
    workers_clean,
    on='picker',
    how='left',
    validate='many_to_one'
)

analysis = analysis.merge(
    zones_clean,
    on='zone',
    how='left',
    validate='many_to_one'
)

print("Combined dataset shape:", analysis.shape)
display(analysis.head())
```

```python

# Copy the original datasets
picks_clean = picks.copy()
workers_clean = workers.copy()
zones_clean = zones.copy()

# Remove duplicate rows
picks_clean = picks_clean.drop_duplicates()
workers_clean = workers_clean.drop_duplicates()
zones_clean = zones_clean.drop_duplicates()

# Handle missing worker experience
workers_clean['experience band'] = (
    workers_clean['experience band'].fillna('Unknown')
)

# Standardize zone names in both datasets
def standardize_zone(series):
    return (
        series.astype('string')
        .str.strip()
        .str.upper()
        .str.replace(r'[\s_-]+', '', regex=True)
        .str.replace('ZONE', 'Zone-', regex=False)
    )

picks_clean['zone'] = standardize_zone(picks_clean['zone'])
zones_clean['zone'] = standardize_zone(zones_clean['zone'])

# Convert timestamps
picks_clean['start'] = pd.to_datetime(
    picks_clean['start'], errors='coerce'
)
picks_clean['finish'] = pd.to_datetime(
    picks_clean['finish'], errors='coerce'
)

# Calculate picking duration in minutes
picks_clean['picking_minutes'] = (
    picks_clean['finish'] - picks_clean['start']
).dt.total_seconds() / 60

# Merge picks with worker details
analysis = picks_clean.merge(
    workers_clean,
    on='picker',
    how='left',
    validate='many_to_one'
)

# Merge zone details
analysis = analysis.merge(
    zones_clean,
    on='zone',
    how='left',
    validate='many_to_one'
)

print("Combined dataset created successfully!")
print("Dataset shape:", analysis.shape)
print("\nColumns:", analysis.columns.tolist())

display(analysis.head())
```

```python

# Step 1: Prepare data for machine learning

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Use the existing feature-engineered dataset
ml_data = feature_data.copy()

# Select available input features
candidate_features = [
    'zone',
    'shift',
    'experience band',
    'quantity',
    'start_hour'
]

input_features = [
    col for col in candidate_features
    if col in ml_data.columns
]

target = 'picking_minutes'

# Keep rows with a valid target
ml_data = ml_data.dropna(subset=[target]).copy()

# Remove invalid negative durations
ml_data = ml_data[ml_data[target] >= 0].copy()

X = ml_data[input_features]
y = ml_data[target]

print("Input features:", input_features)
print("Target variable:", target)
print("Records available:", len(ml_data))
print("Feature matrix shape:", X.shape)
```

```python

# Separate numerical and categorical columns
categorical_features = X.select_dtypes(
    include=['object', 'string', 'category']
).columns.tolist()

numerical_features = X.select_dtypes(
    include=['number']
).columns.tolist()

# Preprocess categorical and numerical data
numeric_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median'))
])

categorical_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

preprocessor = ColumnTransformer([
    ('numeric', numeric_pipeline, numerical_features),
    ('categorical', categorical_pipeline, categorical_features)
])

# Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

# Build and train Random Forest Regressor
model = Pipeline([
    ('preprocessor', preprocessor),
    ('regressor', RandomForestRegressor(
        n_estimators=100,
        random_state=42,
        max_depth=5
    ))
])

model.fit(X_train, y_train)

print("Model training completed!")
print("Training records:", len(X_train))
print("Testing records:", len(X_test))
```

```python

# Predict picking duration for test records
y_pred = model.predict(X_test)

# Calculate evaluation metrics
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("RANDOM FOREST REGRESSION RESULTS")
print("--------------------------------")
print(f"MAE:  {mae:.2f} minutes")
print(f"RMSE: {rmse:.2f} minutes")
print(f"R² Score: {r2:.3f}")
```

```python

plt.figure(figsize=(7, 5))

plt.scatter(y_test, y_pred, alpha=0.7)

# Reference line for perfect predictions
lower = min(y_test.min(), y_pred.min())
upper = max(y_test.max(), y_pred.max())

plt.plot(
    [lower, upper],
    [lower, upper],
    linestyle='--'
)

plt.xlabel('Actual Picking Time (minutes)')
plt.ylabel('Predicted Picking Time (minutes)')
plt.title('Actual vs Predicted Picking Duration')
plt.tight_layout()
plt.show()
```

```python

# Save predictions and actual values
predictions = pd.DataFrame({
    'Actual_Picking_Minutes': y_test.values,
    'Predicted_Picking_Minutes': y_pred
})

predictions.to_csv(
    '/content/round2_model_predictions.csv',
    index=False
)

# Save performance metrics
metrics = pd.DataFrame({
    'Metric': ['MAE', 'RMSE', 'R2 Score'],
    'Value': [mae, rmse, r2]
})

metrics.to_csv(
    '/content/round2_model_metrics.csv',
    index=False
)

print("Model predictions and metrics saved successfully!")
display(metrics)
```
